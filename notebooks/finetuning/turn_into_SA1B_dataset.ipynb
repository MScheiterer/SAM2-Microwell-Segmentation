# SAM2 Finetuning Dataset Creation

This notebook walks through the process of converting a folder of images and the corresponding label-studio ellipse annotations (.json) into the SA-1B format which can be used by the SAM2 Training framework, since Meta provides a Dataloader and Transformer class for finetuning on the SA-1B dataset.

# Usage

Change ```BASE_DIR``` and other paths as needed. If there is an issue with filenames (label studio sometimes changes special characters), check out the function ```remove_commas_percentages(path: str, name: str) -> str``` and adjust its use.

````python
# === Paths ===
BASE_DIR = "../../data/Labelled/chip4_27"
INPUT_IMAGE_DIR = os.path.join(BASE_DIR, "Raw")
INPUT_ANNOTATION_FILE = os.path.join(BASE_DIR, "annotations.json")
OUTPUT_IMAGE_DIR = os.path.join(BASE_DIR, "LikeSA1B", "Images")
OUTPUT_ANNOTATION_DIR = os.path.join(BASE_DIR, "LikeSA1B", "Annotations")

os.makedirs(OUTPUT_IMAGE_DIR, exist_ok=True)
os.makedirs(OUTPUT_ANNOTATION_DIR, exist_ok=True)
````

````python
data = construct_masks_from_json(INPUT_ANNOTATION_FILE)
create_sa1b_dataset(data, INPUT_IMAGE_DIR, OUTPUT_IMAGE_DIR, OUTPUT_ANNOTATION_DIR)
````
