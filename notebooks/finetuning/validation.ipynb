# Validation

This notebook shows how to use the validation framework to validate a set of checkpoints on a specified dataset. See finetuning.validation for further details.

## Import Helper Functions

````python
from finetuning.validation import *
````

## Validation 
Adjust paths as necessary:
- ```configs``` correspond to the names of the finetuned model folders
- ```im_dirs``` and ```ann_dirs``` correspond to the validation dataset (image folder and annotation folder)
- ```output_dirs``` specify where to store the resulting .csv files.

````python
configs = [
    "SAM2-Mask",
    "SAM2-Class",
    "SAM2-IoU",
    "SAM2-Dice"
]

im_dirs = [
    # "../../data/Labelled/finetuning/val/combined/Images",
    # "../../data/Labelled/finetuning/val/chip2/Images",
    "../../data/Labelled/finetuning/val/chip4/Images",
    # "../../data/Labelled/finetuning/val/chip2_preprocessed/Images",
    # "../../data/Labelled/finetuning/val/chip4_preprocessed/Images"
]
ann_dirs = [
    # "../../data/Labelled/finetuning/val/combined/Annotations",
    # "../../data/Labelled/finetuning/val/chip2/Annotations",
    "../../data/Labelled/finetuning/val/chip4/Annotations",
    # "../../data/Labelled/finetuning/val/chip2_preprocessed/Annotations",
    # "../../data/Labelled/finetuning/val/chip4_preprocessed/Annotations"
]


results = []

for training_config in configs:
    output_dirs = [
        # f"./finetuning/{training_config}/combined",
        # f"./finetuning/{training_config}/chip2",
        f"./finetuning/{training_config}/chip4",
        # f"./finetuning/{training_config}/chip2_preprocessed",
        # f"./finetuning/{training_config}/chip4_preprocessed",
    ]
    checkpoint_dir = f"../../models/{training_config}/checkpoints"
    
    for im_dir, ann_dir, output_dir in zip(im_dirs, ann_dirs, output_dirs):
        print(f"""
            Validating {checkpoint_dir.split('/')[-2]} on \n
            Dataset: {im_dir}, \n
            storing in {output_dir}
        """)
        result = validate_checkpoints(
            image_dir=im_dir, 
            ann_dir=ann_dir, 
            checkpoint_dir=checkpoint_dir, 
            output_dir=output_dir
        )
        results.append(result)
````
