# This Notebook shows how to Evaluate a Segmentation Model using the developed Framework. 

Returns:
- Stores a .csv file at ```output_path/experiment_name_summary.csv```

  
Usage: 
1. Set ```experiment_name``` to keep track of which model you're testing
2. Set ```output_path``` to where you want to store the results 
3. Update ```dataset```, or ```im_dir``` and ```ann_dir``` directly as desired
4. Choose sam2 checkpoint and config or other model wrapper

````python
import json
import numpy as np

import os
import sys

from segmentation.segmenter import *
from segmentation.model_evaluation import model_comparer

# Change this
experiment_name = "model_test"

# Update as needed
output_path = "./"
dataset = "test/combined" # or "val"
im_dir = f"../../data/Labelled/finetuning/{dataset}/Images"
ann_dir = f"../../data/Labelled/finetuning/{dataset}/Annotations"

# Define Checkpoints and Configs
checkpoint_IoU = "../../models/SAM2-IoU/checkpoints/checkpoint_230.pt"
checkpoint_Dice = "../../models/SAM2-Dice/checkpoints/checkpoint_280.pt"
checkpoint_Mask = "../../models/SAM2-Mask/checkpoints/checkpoint_230.pt"
checkpoint_Class = "../../models/SAM2-Class/checkpoints/checkpoint_260.pt"

checkpoint_b_plus = "../../sam2/checkpoints/sam2.1_hiera_base_plus.pt"
checkpoint_l = "../../sam2/checkpoints/sam2.1_hiera_large.pt"
checkpoint_s = "../../sam2/checkpoints/sam2.1_hiera_small.pt"
checkpoint_t = "../../sam2/checkpoints/sam2.1_hiera_tiny.pt"

model_cfg_b_plus = "configs/sam2.1/sam2.1_hiera_b+.yaml"
model_cfg_l = "configs/sam2.1/sam2.1_hiera_l.yaml"
model_cfg_s = "configs/sam2.1/sam2.1_hiera_s.yaml"
model_cfg_t = "configs/sam2.1/sam2.1_hiera_t.yaml"

# Choose Model
sam2 = SAM2Wrapper(config=model_cfg_b_plus, checkpoint=checkpoint_IoU, points_per_side=128)
cellposeSAM = CellposeSAMWrapper()

segmenter = Segmenter(model=sam2, filter_config={
        "circularity_threshold": 0.7,
        "border_thres": 2,
        "max_groups": None,
        "eps_area": 40, 
        "eps_coords_medium": 90,
        "eps_coords_small": 70,
        "eps_coords_small_exclude": 20,
        "rough_only": False,
        "debug": True
    })

# Run evaluation
model_comparer(
    image_path=im_dir, 
    annotation_path=ann_dir, 
    segmenter=segmenter, 
    experiment_dir=output_path, 
    experiment_name=experiment_name
)
````
